### Circular Linked List Concepts

**Creating - Linked List**

```python
# With dataclasses
from dataclasses import dataclass
from typing import Optional

@dataclass
class ListNode:
    data: Optional['int'] = 0
    next: Optional['ListNode'] = None
```

```python
def create_circular_ll(array):
    head = ListNode(array[0], None)
    head.next = head

    last = head
    for i in array[1:]:
        t = ListNode(i, None)
        t.next = last.next
        last.next = t
        last = t
    
    return head
```

**Display Circular Linked List**

```python
def display_circular_ll(head):
    if not head:
        print("Empty Linked List")
        
    p = head
    while True:
        print(p.data)
        p = p.next
        if p == head:
            break
```

```python
a = [2,3,4,5,6]
head = create_circular_ll(a)
```

```python
display_circular_ll(head)
```

**Display Circular Linked List - Recursion**

```python
flag = 0
def display_circular_ll_rec(p):
    global head, flag
    if p != head or flag == 0:
        flag = 1
        print(p.data)
        display_circular_ll_rec(p.next)
    
    flag = 0
```

```python
display_circular_ll_rec(head)
```
