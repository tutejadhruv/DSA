### Circular Linked List Concepts

**Creating - Linked List**

```python
# With dataclasses
from dataclasses import dataclass
from typing import Optional

@dataclass
class ListNode:
    data: Optional['int'] = 0
    next: Optional['ListNode'] = None
```

```python
def create_circular_ll(array):
    head = ListNode(array[0], None)
    head.next = head

    last = head
    for i in array[1:]:
        t = ListNode(i, None)
        t.next = last.next
        last.next = t
        last = t
    
    return head
```

**Display Circular Linked List**

```python
def display_circular_ll(head):
    if not head:
        print("Empty Linked List")
        
    p = head
    while True:
        print(p.data)
        p = p.next
        if p == head:
            break
```

```python
a = [2,3,4,5,6]
head = create_circular_ll(a)
```

```python
display_circular_ll(head)
```

**Display Circular Linked List - Recursion**

```python
flag = 0
def display_circular_ll_rec(p):
    global head, flag
    if p != head or flag == 0:
        flag = 1
        print(p.data)
        display_circular_ll_rec(p.next)
    
    flag = 0
```

```python
display_circular_ll_rec(head)
```

**Count number of Elements in Circular Linked List**

```python
def count_circular_ll(head):
    p = head
    count = 0
    while True:
        count += 1
        p = p.next
        if p == head:
            break
    return count
```

```python
a = [2,3,4,5,6]
head = create_circular_ll(a)
count_circular_ll(head)
```

**Inserting in Circular Linked List**

```python
def insert_circular_list(head, index, x):
    length = count_circular_ll(head)
    if index < 0 and index > length:
        return "Invalid Position"
    
    # Inserting at start
    if index == 0:
        t = ListNode(x, None)
        if head == None:
            head = t
            head.next = head
        else:
            p = head
            while True:
                p = p.next
                if p.next == head: 
                    break

            p.next = t
            t.next = head
            head = t # This is optional because circular linked list doesn't have first/ last node.
        return head
    else:
        p = head
        for i in range(index-1):
            p = p.next
        t = ListNode(x, p.next)
        p.next = t
        return head

            
```

```python
a = [2,3,4,5,6]
head = create_circular_ll(a)
head = insert_circular_list(head, 2, 10)
display_circular_ll(head)
```

```python
a = [2,3,4,5,6]
head = create_circular_ll(a)
head = insert_circular_list(head, 5, 10)
display_circular_ll(head)
```
